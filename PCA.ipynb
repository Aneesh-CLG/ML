```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Step 1: Load and standardize the dataset
def load_and_standardize(file_path):
    data = pd.read_csv(file_path)
    numeric_data = data.select_dtypes(include=[np.number])
    standardized_data = (numeric_data - numeric_data.mean()) / numeric_data.std()
    return standardized_data, numeric_data.columns

# Step 2: Perform PCA using sklearn
def perform_pca(data, n_components):
    pca = PCA(n_components=n_components)
    transformed_data = pca.fit_transform(data)
    return pca, transformed_data

# Scree Plot
def scree_plot(pca_model):
    explained_variance_ratio = pca_model.explained_variance_ratio_
    plt.figure(figsize=(8, 6))
    plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio * 100, marker='o', label='Variance Explained (%)')
    plt.xlabel('Principal Components')
    plt.ylabel('Variance Explained (%)')
    plt.title('Scree Plot')
    plt.legend(loc='best')
    plt.grid()
    plt.show()

# Biplot
def biplot(transformed_data, pca_model, feature_names):
    plt.figure(figsize=(8, 6))
    plt.scatter(transformed_data[:, 0], transformed_data[:, 1], alpha=0.6, label='Data Points')
    
    # Plot feature vectors (principal components)
    for i, vector in enumerate(pca_model.components_[:2].T):  # First two principal components
        plt.arrow(0, 0, vector[0], vector[1], color='red', head_width=0.05)
        plt.text(vector[0] * 1.1, vector[1] * 1.1, feature_names[i], color='red')
    
    plt.xlabel('PC1')
    plt.ylabel('PC2')
    plt.title('Biplot')
    plt.legend(loc='upper right')
    plt.grid()
    plt.show()

# Main code
if __name__ == "__main__":
    # Load and standardize the dataset
    file_path = 'Country-data.csv'  # Replace with your dataset path
    standardized_data, feature_names = load_and_standardize(file_path)

    # Perform PCA (using sklearn's PCA)
    n_components = 2  # Number of components to retain
    pca_model, transformed_data = perform_pca(standardized_data, n_components)

    # Generate plots
    scree_plot(pca_model)                    # Scree Plot
    biplot(transformed_data, pca_model, feature_names)  # Biplot
```
