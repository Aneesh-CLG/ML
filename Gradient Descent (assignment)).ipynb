```python
import numpy as np

# Dataset
ages = np.array([41, 42, 37, 46, 48, 44])
heart_rates = np.array([138, 153, 151, 133, 126, 145])
oxycon_actual = np.array([37.99, 47.34, 44.38, 28.17, 27.07, 37.85])

# Initial weights
w = np.array([-59.5, -0.15, 0.6])
learning_rate = 0.000002

# Predictions (Part a)
def predict(w, ages, heart_rates):
    return w[0] + w[1] * ages + w[2] * heart_rates

def compute_sse(predictions, actual):
    return np.sum((predictions - actual) ** 2)

# Step 1: Make predictions
y_pred = predict(w, ages, heart_rates)
print("Predictions:", y_pred)

# Step 2: Compute SSE (Part b)
sse = compute_sse(y_pred, oxycon_actual)
print("Sum of Squared Errors (SSE):", sse)

# Step 3: Gradient Descent Update (Part c)
grad_w0 = 2 * np.sum(y_pred - oxycon_actual)
grad_w1 = 2 * np.sum((y_pred - oxycon_actual) * ages)
grad_w2 = 2 * np.sum((y_pred - oxycon_actual) * heart_rates)

# Update weights
w[0] -= learning_rate * grad_w0
w[1] -= learning_rate * grad_w1
w[2] -= learning_rate * grad_w2
print("Updated Weights:", w)

# Step 4: Compute new SSE (Part d)
y_pred_new = predict(w, ages, heart_rates)
sse_new = compute_sse(y_pred_new, oxycon_actual)
print("New SSE:", sse_new)
```
